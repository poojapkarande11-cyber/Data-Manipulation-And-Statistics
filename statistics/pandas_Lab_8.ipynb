```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('Retail_Dataset_Cleaned.csv')


print('Retail_Dataset_Cleaned.csv')
```

```python
batsman_a = np.array([48,50,49,51,52,50])
batsman_b = np.array([10,90,30,70.100,0])

print("A -> Variance:", np.var(batsman_a), "Std Dev:", np.std(batsman_a))
print("B -> Variance:", np.var(batsman_b), "Std Dev:", np.std(batsman_b))
```

```python
## Calculate Revenue Skewness

revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

```python
## Interpret Revenue Skewness

if revenue_skew > 0:
    print("Positive Skew")
elif revenue_skew < 0:
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

```python

## Calculate Units Sold Skewness

units_skew = df["Units_Sold"].skew()
print(units_skew)
```

```python

## Task 2: Understanding Distribution Patterns

import matplotlib.pyplot as plt
```

```python

## Plot Revenue Distribution

plt.hist(df["Revenue"], bins=20)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python
## Plot Units Sold Distribution

plt.hist(df["Units_Sold"], bins =20)
plt.title("Units Sold Distribution")
plt.xlabel("Units Sold")
plt.ylabel("Frequency")
plt.show()
```

```python
## Display Final Summary Table

summary = pd.DataFrame({
    "Metric":["Revenue","Units Sold"],
    "Skewness":[revenue_skew, units_skew]
})
summary
```
