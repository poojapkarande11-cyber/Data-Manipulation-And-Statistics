```python
## Task 2: Filter Data

## Filter Products with Unit Price Greater Than 1000

high_price = df[df['Unit_Price'] > 1000]
high_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
## Filter Products with Unit Price Less Than 500

low_price = df[df['Unit_Price'] < 500]
low_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
## Filter Products with Unit Price Between 500 and 1500

medium_price = df[
    (df['Unit_Price'] >= 500) &
    (df['Unit_Price'] <= 1500)
]
medium_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python
## Sort Data by Unit Price (Highest to Lowest)

sorted_price = df.sort_values(
    by='Unit_Price',
    ascending=False
)
sorted_price[['Product_Name', 'Category', 'Unit_Price']].head()
```

```python

# Display Top 10 Highest Unit Price Products

top_price = df.nlargest(10, 'Unit_Price')
print(top_price[['Unit_Price']])
```
